```python
import rasterio
import csv
from datetime import datetime, timedelta
```

```python
file = r"C:/Users/Carol/Dropbox/GIS/Projects/RFO_mining/2_intermediate_data/raster/GFW_RFO_clip_20241023.tif"
```

```python
# Open the raster file
with rasterio.open(file) as src:
    # Get the geospatial information
    transform = src.transform
    x_size = transform[0]
    y_size = -transform[4]
    x_min = transform[2]
    y_max = transform[5]
    # Loop through the pixels
    for i in range(src.width):
        for j in range(src.height):
            # Get the pixel value
            pixel_value = str(src.read(1)[j][i])
            # Check if the pixel value is not 0
            if pixel_value != '0':
                # Extract the confidence level and date
                confidence_level = pixel_value[0]
                date_str = pixel_value[1:]
                # Convert the date to a datetime object
                date = datetime(2014, 12, 31) + timedelta(int(date_str))
                # Get the longitude and latitude
                longitude = x_min + (i + 0.5) * x_size
                latitude = y_max - (j + 0.5) * y_size
                # Write to the CSV file
                with open('GFW_20241023.csv', 'a', newline='') as f:
                    writer = csv.writer(f)
                    writer.writerow([longitude, latitude, date.strftime('%Y-%m-%d'), confidence_level])
```

```python
# Initialize a variable to hold the total deforestation area
total_deforestation_area = 0

# Loop through the pixels
for i in range(src.width):
    for j in range(src.height):
        # Get the pixel value
        pixel_value = str(src.read(1)[j][i])
        # Check if the pixel value is not 0 and confidence level starts with 3 or 4
        if pixel_value != '0' and pixel_value[0] in ['3', '4']:
            # Extract the confidence level and date
            confidence_level = pixel_value[0]
            date_str = pixel_value[1:]
            # Convert the date to a datetime object
            date = datetime(2014, 12, 31) + timedelta(int(date_str))
            # Get the longitude and latitude
            longitude = x_min + (i + 0.5) * x_size
            latitude = y_max - (j + 0.5) * y_size
            # Calculate the deforestation area (10m x 10m pixels)
            deforestation_area = 10 * 10  # Pixel area in square meters
            # Add the deforestation area to the total
            total_deforestation_area += deforestation_area

print(f"Total deforestation area with confidence level 3 or 4: {total_deforestation_area} square meters")
```
